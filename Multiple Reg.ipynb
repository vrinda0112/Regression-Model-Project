```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn import linear_model
```

```python
df = pd.read_csv("car data.csv")
```

```python
df.head()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df.describe()
```

```python
df.columns
```

```python
X = df[['Present_Price', 'Year', 'Kms_Driven',
        'Fuel_Type', 'Seller_Type', 'Transmission', 'Owner']]

Y = df['Selling_Price']
```

```python
X = pd.get_dummies(
    X,
    columns=['Fuel_Type', 'Seller_Type', 'Transmission'],
    drop_first=True,
    dtype=int
)
```

```python
X.head()
```

```python
X.info()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42
)
```

```python
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
reg = linear_model.LinearRegression()

reg.fit(X_train, Y_train)
```

```python
print("Coefficients:", reg.coef_)
print("Intercept:", reg.intercept_)
```

```python
Y_pred = reg.predict(X_test)

print(Y_pred[:10])
```

```python
comparison = pd.DataFrame({
    'Actual Selling Price': Y_test.values,
    'Predicted Selling Price': Y_pred
})

comparison.head(10)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

mae = mean_absolute_error(Y_test, Y_pred)
mse = mean_squared_error(Y_test, Y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(Y_test, Y_pred)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R2 Score:", r2)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(Y_test, Y_pred, color='blue')

plt.xlabel("Actual Selling Price")
plt.ylabel("Predicted Selling Price")
plt.title("Actual vs Predicted Selling Prices")

plt.show()
```

```python
new_car = pd.DataFrame({
    'Present_Price': [8.0],
    'Year': [2018],
    'Kms_Driven': [40000],
    'Owner': [0],
    'Fuel_Type_Diesel': [1],
    'Fuel_Type_Petrol': [0],
    'Seller_Type_Individual': [1],
    'Transmission_Manual': [1]
})

predicted_price = reg.predict(new_car)

print("Predicted Selling Price:", predicted_price[0], "Lakhs")
```
